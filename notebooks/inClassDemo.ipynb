```python

import numpy as np
import asyncio
import math

# ---------- helper utilities ----------
def wrap_to_pi(angle):
    return (angle + np.pi) % (2 * np.pi) - np.pi

async def stop_robot():
    cmdr.set_vel(0.0, 0.0)
    await asyncio.sleep(0.1)

async def log_and_plot_step(odom_log, gt_log, dt=0.05):
    pose, gt_pose = cmdr.get_pose()
    odom_log.append(np.array(pose))
    gt_log.append(np.array(gt_pose))
    cmdr.plot_odom(pose[0], pose[1])
    cmdr.plot_gt(gt_pose[0], gt_pose[1])
    await asyncio.sleep(dt)

async def move_forward_for(duration, linear_speed, odom_log, gt_log, dt=0.05):
    t = 0.0
    while t < duration and cmdr.sim_is_running() and cmdr.plotter_is_running():
        cmdr.set_vel(linear_speed, 0.0)
        await log_and_plot_step(odom_log, gt_log, dt)
        t += dt
    await stop_robot()

async def turn_in_place_for(duration, angular_speed, odom_log, gt_log, dt=0.05):
    t = 0.0
    while t < duration and cmdr.sim_is_running() and cmdr.plotter_is_running():
        cmdr.set_vel(0.0, angular_speed)
        await log_and_plot_step(odom_log, gt_log, dt)
        t += dt
    await stop_robot()
```


# Completed Task 1: Open-Loop Square + Task 2: Closed-Loop Obstacle Avoidance

These cells provide a working baseline for the in-class tasks:

1. **Open-loop square**: drive forward and turn 90° four times, while plotting odometry and ground truth.
2. **Closed-loop obstacle avoidance**: move forward until the robot gets too close to an obstacle, then turn away.

Run the setup cells first so the simulator and plotter are already launched.


## Task 1: Open-loop square

Tune the two main parameters:
- `FWD_TIME`: how long each straight segment lasts
- `TURN_TIME`: how long the 90° turn lasts

If your square is too small or too large, change `FWD_TIME`.
If the corners are not close to 90°, change `TURN_TIME`.


```python
# ---------- open-loop square ----------
cmdr.reset_plotter()
cmdr.reset_sim()
cmdr.plot_map()

odom_log = []
gt_log = []

# Baseline parameters to start with
LINEAR_SPEED = 0.25      # m/s
ANGULAR_SPEED = 1.0      # rad/s
FWD_TIME = 0.9           # seconds
TURN_TIME = (np.pi / 2) / ANGULAR_SPEED *0.605  # ideal time for ~90 deg turn

for _ in range(4):
    await move_forward_for(FWD_TIME, LINEAR_SPEED, odom_log, gt_log)
    await turn_in_place_for(TURN_TIME, ANGULAR_SPEED, odom_log, gt_log)

await stop_robot()

odom_log = np.array(odom_log)
gt_log = np.array(gt_log)

print(f"Collected {len(odom_log)} pose samples.")
if len(odom_log) > 0:
    odom_closure_error = np.linalg.norm(odom_log[-1, :2] - odom_log[0, :2])
    gt_closure_error = np.linalg.norm(gt_log[-1, :2] - gt_log[0, :2])
    print(f"Odometry closure error: {odom_closure_error:.3f} m")
    print(f"Ground-truth closure error: {gt_closure_error:.3f} m")
```


## Task 2: Closed-loop obstacle avoidance

This controller:
- moves forward while there is enough clearance
- turns in place when the robot gets too close to an obstacle

It uses the **minimum** sensor reading as the safety signal, which makes it robust even if there are multiple range sensors.


```python

# ---------- closed-loop obstacle avoidance ----------
cmdr.reset_plotter()
cmdr.reset_sim()
cmdr.plot_map()

odom_log2 = []
gt_log2 = []

FORWARD_SPEED = 0.25     # m/s
TURN_SPEED = 1.2         # rad/s
SAFE_DIST = 0.40         # m, increase if you still collide
TURN_TIME_AVOID = 0.9    # s, increase if the robot does not turn enough
RUN_TIME = 30.0          # total experiment duration
DT = 0.05

t = 0.0
while t < RUN_TIME and cmdr.sim_is_running() and cmdr.plotter_is_running():
    sensor_values = np.array(cmdr.get_sensor()).astype(float).flatten()
    min_dist = float(np.min(sensor_values))

    if min_dist < SAFE_DIST:
        # Too close: turn in place to avoid collision
        cmdr.set_vel(0.0, TURN_SPEED)
        turn_elapsed = 0.0
        while turn_elapsed < TURN_TIME_AVOID and cmdr.sim_is_running() and cmdr.plotter_is_running():
            await log_and_plot_step(odom_log2, gt_log2, DT)
            turn_elapsed += DT
            t += DT
    else:
        # Safe: continue forward
        cmdr.set_vel(FORWARD_SPEED, 0.0)
        await log_and_plot_step(odom_log2, gt_log2, DT)
        t += DT

await stop_robot()

odom_log2 = np.array(odom_log2)
gt_log2 = np.array(gt_log2)

print(f"Collected {len(odom_log2)} pose samples for obstacle avoidance.")
```


## What to write in your answers

### Open loop
- A velocity command lasts for the amount of time you choose in the code (`FWD_TIME` or `TURN_TIME`).
- The robot will usually **not** trace the exact same square every time.
- Odometry and ground truth will generally drift apart over time.

### Closed loop
- The turn amount depends on your turn speed and turn duration.
- Faster forward speed makes collisions more likely.
- A larger `SAFE_DIST` is safer, but may make the robot turn too early.
- If collisions still happen, try:
  - increasing `SAFE_DIST`
  - lowering `FORWARD_SPEED`
  - increasing `TURN_TIME_AVOID`
